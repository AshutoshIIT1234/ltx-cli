# ⚡ LTX CLI — GPU Test Notebook

This notebook tests the LTX CLI on Google Colab's free T4 GPU (15GB VRAM).

**Steps:**
1. Install LTX CLI from GitHub (handling dependencies)
2. Verify GPU detection
3. Run dry-run tests
4. Generate actual video

> Make sure to select **Runtime → Change runtime type → T4 GPU** before running.

## Step 1: Install LTX CLI from GitHub

## Step 4: Dry Run (no GPU usage)

```
!ltx generate --dry-run --preset balanced --prompt "cinematic shot of rain in tokyo at night"
```

## Step 5: 🎬 Generate Actual Video!

This will:
1. Download the LTX-Video model (~9GB, first time only)
2. Generate video frames on the T4 GPU
3. Save the output as MP4

**Using `fast` preset** since Colab T4 has ~15GB VRAM.

```
!ltx generate --preset fast --prompt "A cinematic robot walking in the rain at night, neon lights reflecting on wet streets" --output ./outputs/robot_rain.mp4
```

## Step 6: View the Output

```
from IPython.display import Video, display
import os

video_path = './outputs/robot_rain.mp4'
if os.path.exists(video_path):
    display(Video(video_path, embed=True, width=640))
    print('\n✅ Video generated successfully!')
else:
    print('❌ Video file not found. Check the generation output above for errors.')
```
